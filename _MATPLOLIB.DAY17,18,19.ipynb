```python
plt.figure(figsize = (15,5), dpi=200)
plt.plot(x,x**2,'r--*')
plt.plot(x,x**3,'m-->')

plt.xticks(range(0,31,1))
plt.yticks(range(0,27001,3000))

plt.title('Square vs Cube')
plt.xlabel('Numbers')
plt.ylabel('Square/Cube')
```

```python
plt.figure(figsize = (15,5), dpi=200)
plt.plot(x,x**2,'r--*',label = 'Square')
plt.plot(x,x**3,'m-->',label = 'Cube')

plt.xticks(range(0,31,1))
plt.yticks(range(0,27001,3000))

plt.title('Square vs Cube')
plt.xlabel('Numbers')
plt.ylabel('Square/Cube')
plt.legend()

plt.savefig('Square vs Cube',dpi=200)
```

## SCATTER PLOT

```python
plt.figure(dpi=250)
plt.scatter(x,y,c='m',s=20)
```

## BARPLOT

```python
x = np.arange(10)
y = x*2

plt.figure(dpi=150)
plt.bar(x,y)
```

```python
x = np.arange(10)
y = x*2

plt.figure(dpi=150)
plt.bar(x,y,width=.5,color='y')
```

## BOX PLOT

```python
x= np.random.randint(1,100,10000)
plt.figure(dpi=100)
plt.boxplot(x)
plt.show()
```

## PIE CHART

```python
x =  np.array([10,20,30,15,40])
plt.pie(x)
plt.show()
```

```python
x =  np.array([10,20,30,15,40])

plt.figure(figsize = (15,5), dpi=100)
fruits = ['Apple','Orange','Mango','Kiwi','Dragon fruit']
plt.pie(x,labels=fruits, autopct='%1.2f%%')
plt.show()
```
