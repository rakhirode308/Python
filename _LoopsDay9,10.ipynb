## BREAK
- USED TO TERMINATE THE LOOP

```python
l = [11,23,45,88,66,67,98,78,25]
count = 0
num = int(input('Enter any integer :'))

for i in l:
    count = count+1
    if i==num:
        print('The number is present in list')
print('Number of iterations:',count)
```

```python
l = [11,23,45,88,66,67,98,78,25]
count = 0
num = int(input('Enter any integer :'))

for i in l:
    count = count+1
    if i==num:
        print('The number is present in list')
        break
print('Number of iterations:',count)
```

```python
l = [11,23,45,88,66,67,98,78,25]

num = int(input('Enter any integer :'))

for i in l:
    count = count+1
    if i==num:
        print('The number is present in list')
        break
else:
    print('The number is not present in list')
```

```python
l = [11,11,33,33,44,33,11,33,55,55]
Ques: Given a list of numbers , find the frequency of occurence of distinct numers.

11 OCCURED 3 TIMES
33 OCCURED 4 TIMES
44 OCCURED 1 TIMES
55 OCCURED 2 TIMES
```

```python
l = [11,11,33,33,44,33,11,33,55,55]
fre = {}
for i in l:
    if i in fre.keys():
        fre[i] += 1
    else:
        fre[i] = 1
for i,j in fre.items():
    print(i,'OCCURED',j,'TIMES')
```

```python
l = [11,11,33,33,44,33,11,33,55,55]

largest_no = l[0]
for i in l:
    if i > largest_no:
        largest_no = i
print('The largest number in the list is :', largest_no)
```
