```python
# INSERT 
''' np.insert(array,index,value,axis=None)
    array - original array
    index - position of new value to be inserted
    value - actual data or new value
    axis :- 0, data to be inserted row-wise, for 2D array,
    axis :- 1, column-wise insert
'''  
import numpy as np

arr = np.array([10,20,30,40,50,60])
print(arr)
new_arr = np.insert(arr,2,100)
print(new_arr)
```

```python
#BROADCASTING
'''
Broadcasting is a numpy way where you perform operations on various shapes of array,
without using loops.
'''
```

```python
# WITH LOOP:
prices = [100,200,300]
discount = 10
final_prices = []

for price in prices:
    final_price = price - (price * discount/100)
    final_prices.append(final_price)
print(final_prices)
```

```python
# WITHOUT LOOPS: 
import numpy as np
prices = np.array([100,200,300])
discount = 10 # scalar single value
final_prices = prices-(prices * discount/100)

print(final_prices)
```

```python
# HANDLING MISSING VALUES
'''
1. np.isnan() - detects missing values
2. np.nan_to_num() - to replace missing values from nan to other value
3. np.isinf() - to detect infinite values
'''
```

```python
# NAN - NOT A NUMBER, returns Booleaan values
```

```python
import numpy as np
arr = np.array([1,2,np.nan,4,np.nan,6])
print(np.isnan(arr))
```

```python
import numpy as np
arr = np.array([1,2,np.nan,4,np.nan,6])
print(np.isnan(arr))

print(np.nan == np.nan)
```

```python
# np.nan cannot be compared directly
```

```python
#np.nan_to_num(array, nan=value) default=0
# returns a new array

import numpy as np
arr = np.array([1,2,np.nan,4,np.nan,6])
cleaned_arr = np.nan_to_num(arr, nan=100)
print(cleaned_arr)
```

```python
#np.isinf() , returns boolean value.
#1/0
import numpy as np
arr = np.array([1,2,np.inf,4,-np.inf,6])
print(arr)
```

```python
#np.isinf() 
#1/0
import numpy as np
arr = np.array([1,2,np.inf,4,-np.inf,6])
print(np.isinf(arr))
```

```python
# Replace Infinite Values with Finite Values
import numpy as np
arr = np.array([1,2,np.inf,4,-np.inf,6])
print(np.isinf(arr))
cleaned_arr = np.nan_to_num(arr, posinf=1000, neginf=1000)
print(cleaned_arr)
```
