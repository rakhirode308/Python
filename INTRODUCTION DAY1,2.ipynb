# Write a program to Swap two numbers given by User

```python
num1=int(input("Enter first number"))
num2=int(input("Enter second number"))
temp=num1
num1=num2
num2=temp
print("\nAfter swapping first number is :", num1)
print("After swapping first number is :", num2)
```

```python
num1=int(input("Enter first number"))
num2=int(input("Enter second number"))
num1=num1+num2
num2=num1-num2
num1=num1-num2
print("\nAfter swapping first number is :", num1)
print("After swapping first number is :", num2)
```

```python
num1=int(input("Enter first number"))
num2=int(input("Enter second number"))
num1,num2=num2,num1
print("\nAfter swapping first number is :", num1)
print("After swapping first number is :", num2)
```

# Write a Program to calculate The Simple Interest and Total Amount
- S.I. = Amount*Rate*Time/100
- T.A. = Amount+S.I.

```python
Amt=int(input("Enter Amount"))
Rate=int(input("Enter Rate"))
Time=int(input("Enter Time"))
SI=Amt*Rate*Time/100
TA=Amt+SI
print("\nSimple Interst is :", SI)
print("Total Amount at :", Rate, "% rate of interest after", Time,"years is :",TA)
```

```python
Amt=int(input("Enter Amount"))
Rate=int(input("Enter Rate"))
Time=int(input("Enter Time"))
SI=Amt*Rate*Time/100
TA=Amt+SI
print("\nSimple Interst is :", SI)
print("Total Amount at {} % rate of interest after {} years is" .format(Rate,Time),TA)
```
