## ENCAPSULATION

    EX: SPLIT FUNCTION

```python
class Person:
    def __init__ (self,name,age,city):               # init is a special function or constructor used to construct the data of the class 
        self.name = name
        self.age = age
        self.city = city
    def show_details(self):
        print('\nThe Details are :')
        print('Name :',self.name)
        print('Age :',self.age)
        print('City :',self.city)
```

```python
class Person2:
    def __init__ (self):
        self.name = None
        self.age = None
        self.city = None
    def get_details(self,name,age,city):
        self.name = name
        self.age = age
        self.city = city
    def show_details(self):
        print('\nThe Details are :')
        print('Name :',self.name)
        print('Age :',self.age)
        print('City :',self.city)
    def update_city(self,new_city):
        self.city = new_city
        print('City for {} has been changed to {}'.format(self.name, self.city))
        
```

```python
p1 = Person2()
```

```python
p1.name
```

```python
p1.show_details()
```

```python
p1.get_details('Disha',23,'pune')
```

```python
p1.name
```

```python
p1.show_details()
```

```python
p1.update_city('Delhi')
```

```python
p1.show_details()
```

## INHERITANCE
- IT MEANS ONE CLASS(CHILD) CAN USE THE PROPERTIES AND METHOD FROM ANOTHER CLASS(PARENT).
- SIMPY MEANS ACQUIRING PROPERTIES FROM ANOTHER CLASS.

```python
class Employee(Person):
    pass
```

```python
e1 = Employee('Rakhi',23,'pune')
```

```python
e1.name
```

```python
e1.show_details()
```

```python
class Ethans:
    
    def __init__(self, course, duration):
        self.course = course
        self.duration = duration
   
    def show_course(self):
        print("Course:", self.course)
    
    def show_duration(self):
        print("Duration:", self.duration)
    
    def change_course(self, new_course):
        self.course = new_course
        print("Course changed to:", self.course)
    
    def display_info(self):
        print("Course:", self.course)
        print("Duration:", self.duration)



e = Ethans("Python", "6 Months")


e.show_course()
e.show_duration()
e.change_course("Data Science")
e.display_info()
```
