##  IF ELIF ELSE
    ELIF IS USED WHEN MULTIPLE CONDITIONS ARE THERE.
    ONLY ONE IF AND ONE ELSE SHOULD BE THERE.
    THERE CAN BE MULTIPLE ELIF.

```python
a = int(input('Enter first number:'))          # perform add subt 
b = int(input('Enter second number:'))
op = input('Enter the operation(+/-):')

if op == '+':
    print('The addition of', a, 'and', b, 'is:' ,a+b)
elif op == '-':
    print('The subtraction of', a, 'and', b, 'is:' ,a-b)
else:
    print('Operator other than + and - are not allowed')
```

- Write a program for mini calculator that can calculate (+,-,*,/,//,**).
- Write a program to find the largest, second largest and third largest number when three number are provided by the user.

```python
a = int(input('Enter first number:'))         
b = int(input('Enter second number:'))
op = input('Enter the operation(+,-,/,*,//,**):')

if op == '+':
    print('The addition of', a, 'and', b, 'is:' ,a+b)
elif op == '-':
    print('The subtraction of', a, 'and', b, 'is:' ,a-b)
elif op == '/':
    print('The division of', a, 'and', b, 'is:' ,a/b)
elif op == '*':
    print('The multiplication of', a, 'and', b, 'is:' ,a*b)
elif op == '//':
    print('The floor division of', a, 'and', b, 'is:' ,a//b)
elif op == '**':
    print('The exponentiation of', a, 'and', b, 'is:' ,a**b)
else:
    print('Invalid Operator')
```

### NESTED IF ELSE CONDITION

```python
username = input('Enter the username:')
password = input('Enter the password:')

if username =='rakhi383' and password =='12345':
    print('ACCESS GRANTED')
else:
    if username != 'rakhi383':
        print('username is incorrect')
    else:
        print('password is incorrect')
```

```python
a = int(input('Enter first number:'))         
b = int(input('Enter second number:'))
c = int(input('Enter third number:'))

if a>b and a>c:
    if b>c:
        print(b, 'is second largest number')
    else:
        print(c, 'is a second largest number')
elif b>c:
    if a>c:
        print(a, 'is second largest number')
    else:
        print(c, 'is a second largest number')
else:
    if a>b:
        print(a, 'is second largest number')
    else:
        print(b, 'is second largest number')
```

```python
## Second largest number
a = int(input('Enter first number:'))         
b = int(input('Enter second number:'))
c = int(input('Enter third number:'))

if (a>b and a<c) or (a<b and a>c):
    print(a, 'is second largest number')
elif (b<a and b>c) or (b<c and b>a):
    print(b, 'is second largest number')
else:
    print(c, 'is second largest number')
```
